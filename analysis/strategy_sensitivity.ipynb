# Thief strategy sensitivity
One-factor sweep on a fixed dangerous state.

```python
import json
from pathlib import Path

rows = json.loads(Path(
    '../artifacts/analysis/strategy-sensitivity.json'
).read_text())
print('weight | move | risk   | total    | margin')
for row in rows:
    print(f"{row['capture_risk_weight']:>3} | "
          f"{row['chosen_move']:^4} | {row['chosen_risk']!s:>6} | "
          f"{row['chosen_total']!s:>8} | {row['margin_to_second']}")
```

The selected move remains stable while its safety margin changes. This checks robustness; it does not estimate win rate.
